```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import SGDRegressor

#training data
x = np.array([0, 1, 2, 3, 4, 5])
y = np.array([0, 1, 5, 8, 9, 12])

plt.scatter(x, y)
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

```python
X = x.reshape(-1, 1) # Scikit learn expects a 2d array
print("X: ", X)

model = SGDRegressor(
    penalty=None,
    learning_rate="constant",
    eta0=0.01,
    random_state=13
)

costs = []

for epoch in range(500):
  model.partial_fit(X, y)
  y_hat = model.predict(X)
  cost = np.mean((y_hat - y) **2)
  costs.append(cost)

print("Weight (w): ", model.coef_[0])
print("Bias (b): ", model.intercept_[0])
print(costs[:3])
```

```python
# plot 1: training data v predictions
plt.scatter(x, y, label="Actual")
plt.plot(x, model.predict(X), label="Predicted")
plt.legend()
plt.show()
```

```python
#plot 2: cost v training epochs
plt.plot(costs[:])
plt.xlabel("Epoch")
plt.ylabel("Mean Squared Error")
plt.show()
```
